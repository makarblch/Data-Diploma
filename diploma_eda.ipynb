### EDA

1. Гистограмма таргета

- 40% выборки имеют нулевой оборот
- довольно

```python
import matplotlib.pyplot as plt

zero_cnt = (data[target] == 0).sum()
non_zero = data[data[target] >= 0][target]
p98 = data[target].quantile(0.99)

plt.figure(figsize=(12,6))

plt.hist(non_zero[non_zero <= p98], bins=100)

plt.axvline(
    0,
    color='red',
    linestyle='--',
    label=f'Zero values: {zero_cnt:,}'
)

plt.title("Target distribution without zero values")
plt.xlabel(target)
plt.ylabel("Count")

plt.legend()

plt.show()
```

2. Гистограмма без нулей

```python
non_zero = data[(data[target] > 0) & (data[target] <= p98)]

plt.figure(figsize=(12,6))

plt.hist(
    non_zero[target],
    bins=100
)

plt.title("Non-zero target distribution")

plt.show()
```

3. Гистограмма величины = log(target) без нулей.
Нормальное распределение

```python
import numpy as np

plt.figure(figsize=(12,6))

plt.hist(
    np.log1p(non_zero[target]),
    bins=100
)

plt.title("Log-transformed non-zero target")

plt.show()
```

#### Выводы по анализу таргета:

1. Таргет имеет extremely heavy-tailed distribution - очень много нулевых значений, причем природа нуля не до конца понятна: либо отсутствие реального оборота, либо недостаточно хорошее покрытие источниками данных об обороте.

2. Классическая MSE-регрессия скорее всего будет нестабильна. Нужно пробовать подходы с робастными методами

3. Необходимо убирать выбросы - некоторые значения с крайне высокими оборотами сильно смещают оценки

4. Таргет имеет Log-normal распределение - для решения задачи будем использовать именно это

#### Кластеризация по ОКВЭД

```python
data["okved_group"] = (
    data["main_okved"]
    .astype(str)
    .str.split(".")
    .str[0]
)
```

```python
upper_target = data["max_turnover"].quantile(0.995)

data["max_turnover_clipped"] = (
    data["max_turnover"]
    .clip(upper=upper_target)
)

data["log_target"] = np.log1p(
    data["max_turnover_clipped"]
)
```

```python
import pandas as pd
pd.set_option('display.max_columns', None)

okved_stats = (
    data
    .groupby("okved_group")
    .agg({
        "log_target": [
            "median",
            "mean",
            "std"
        ],

        "log_income": [
            "median"
        ],

        "transaction_amt_is_null": [
            "mean"
        ],

        "ip_flg": [
            "mean"
        ]
    })
)

okved_stats
```

```python
okved_stats.columns = [
    "_".join(col)
    for col in okved_stats.columns
]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_okved = scaler.fit_transform(okved_stats)
```

```python
from sklearn.cluster import AgglomerativeClustering

cluster_model = AgglomerativeClustering(
    n_clusters=5
)

okved_stats["okved_cluster"] = (
    cluster_model.fit_predict(X_okved)
)
```

```python
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA

# PCA до 2 компонент
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_okved)

# plotting
plt.figure(figsize=(10, 8))

scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=okved_stats["okved_cluster"],
    alpha=0.8
)

plt.xlabel("PCA 1")
plt.ylabel("PCA 2")

plt.title("OKVED Behavioral Clusters")

plt.colorbar(scatter)

plt.show()
```

```python
import seaborn as sns

cluster_profiles = (
    okved_stats
    .groupby("okved_cluster")
    .mean()
)

plt.figure(figsize=(12,8))

sns.heatmap(
    cluster_profiles,
    annot=True,
    cmap="coolwarm"
)

plt.title("Behavioral Profiles of OKVED Clusters")

plt.show()
```

Кластеры в первую очередь разделились по:

1. размеру бизнеса;
2. coverage transaction data;
3. доле ИП;
4. стабильности оборота.

```python
okved_stats["okved_cluster"].value_counts()
```

```python
okved_stats[
    okved_stats["okved_cluster"] == 4
]
```

```python
# Добавляем кластеры
# маппинг okved_group -> okved_cluster

okved_cluster_map = okved_stats["okved_cluster"]

data["okved_cluster"] = (

    data["okved_group"]

    .map(okved_cluster_map)

)

# если какие-то okved_group не попали в okved_stats

data["okved_cluster"] = data["okved_cluster"].fillna(-1).astype(int)
```

### Формирование модельного датасета

```python
import numpy as np
import pandas as pd

clip_q = 0.995
target_raw = "max_turnover"

# target clipping
upper_target = data[target_raw].quantile(clip_q)

data["max_turnover_clipped"] = data[target_raw].clip(upper=upper_target)
data["log_target"] = np.log1p(data["max_turnover_clipped"])

# если log_income еще нет
data["log_income"] = np.log1p(data["finalincome"])

# buckets
data["log_income_bucket"] = pd.qcut(
    data["log_income"],
    q=10,
    duplicates="drop"
)

data["company_age_bucket"] = pd.cut(
    data["company_age"],
    bins=[-1, 12, 36, 84, 180, np.inf],
    labels=["0-12", "13-36", "37-84", "85-180", "180+"]
)

data["has_transaction_data"] = 1 - data["transaction_amt_is_null"]

# признаки для ML
ml_features = [
    "log_income",
    "finalincome",

    "all_bki_current_pmt",
    "all_tinkoff_current_pmt",
    "payment_to_income",
    "payment_bki_to_income",
    "payment_tkf_to_income",

    "transaction_amt",
    "transaction_cnt",
    "pred_transaction_amt",
    "avg_transaction",
    "business_signal",
    "transactions_to_income",

    "company_age",
    "ip_flg",

    "transaction_amt_is_null",
    "transaction_cnt_is_null",
    "pred_transaction_amt_is_null",
    "business_signal_is_null",
    "transactions_to_income_is_null",
    "income_x_transaction_is_null",
    "annual_revenue_is_null",
    "vital_business_flg_is_null",

    "risk_flg",
    "phoenix_reject_flg",
    "fns_status_problems_flg",
    "vital_business_flg",

    "okved_cluster"
]

segment_cols = [
    "log_income_bucket",
    "ip_flg",
    "company_age_bucket",
    "transaction_amt_is_null",
    "okved_cluster"
]

service_cols = [
    "secret_id",
    "rn",
    "created_dt",
    "max_turnover",
    "max_turnover_clipped",
    "log_target"
]
all_model_cols = (
    service_cols
    + ml_features
    + segment_cols
)

# удаляем дубли, сохраняя порядок
all_model_cols = list(dict.fromkeys(all_model_cols))

model_data = data[all_model_cols].copy()

print(model_data.shape)
print(model_data.columns.tolist())
```

```python
bucket_cols = [
    "log_income_bucket",
    "company_age_bucket"
]

for col in bucket_cols:
    model_data[col] = model_data[col].astype(str)
```

```python
# сохраняем в паркет - для удобства хранения и чтения
model_data.to_parquet("model_data_extended_2.parquet", index=False)
```
